### Section 1. Basic Concepts

__print()__

A function that simply prints something in the screen. In below snippet, computer prints 'Hello world!' in the screen.

```python
print('Hello world!') 
```

```python
print("Hello world double quotation!")
print('Hello world single quotation!')
```

```python
print("I'm a student")
```

```python
print('Hello, World!', end = ' ')
print('Python speaking!')
```

__Variables__ 

A variable has a name and can be used to hold values. 

A variable name must:
1. start with a letter or underscore (_),
2. contains only alpha-numeric characters and underscores,
3. not be Pyhon keywords, such as 'if', 'else', 'while', etc.

Examples of acceptable variable names: 'a_number', '_number', 'number1'.
Examples of unaccetable variable names: '1_number', 'number$', 'if'.

Variable names are case sensitive. For example, "a_number" and "A_Number" are two different variables.

There is no strict requirement for variable naming convention. It is preferred to use the "snake case" style, e.g., lowercase workds connected by under scores. For example, "a_good_variable_name".

__Order of Operators__

In Python, order of operators is just like it is in mathematics.
1) power operator ("**")
2) multiplation and division (standard division, "/", integer division "//", modulus division, "%")
3) addition ("+") and subtraction ("-")

Like in mathematics, we can use bracket to change the order.

Note a special case, if we use multiple power operations, Python operates from right to left.

```python
2 ** 2 ** 3 # = 2 ** (2 ** 3) = 2 ** 8 = 256
```

__Bit Operators__ (advanced concept, good to know but not required.)


```python
first_bit = 1
second_bit = 0

# '&' (and): both two bits are 1 then result is 1, otherwise, 0.
print(first_bit & second_bit) # 0

# '|' (or): either bit is 1 then 1.
print(first_bit | second_bit) # 1

# '^' (xor): if the two bits are different then 1.
print(first_bit ^ second_bit) # 1

# '~' (bitwise not): x = -x - 1
print(~1) # -2
print(~0) # -1
# For example, convert number to signed binary, 5 = '0000 0101'. Bitwise not of 5 becomes '1111 1010' (-6)

# '<<' (left shift, equivalent to multiplying the first number by 2 powers of the second number)
print(2 << 1) # 4: 2 = '0010', shift left 1 bit, '0100' (4), equivalent to 2 * 2 = 4
print(5 << 1) # 6: 5 = '0101', shift left 1 bit, '1010' (10), equivalent to 5 * 2 = 10

# '>>' (right shift, equivalent to dividing the first number by 2 powers of the second number)
print(4 >> 2) # 1, 4 = '0100', shift right 2 bits, '0001' (1), equivalent to 4 / (2**2) = 1
```

__Joining Multiple Conditions__

Use "and", "or", "not". Priorities are "not" > "and" > "or".

```python
user_age = 14
gender = "F"
if user_age < 18 and gender == "F":
    print("You are a girl")
elif user_age < 18 and gender == "M":
    print("You are a boy")
elif user_age >= 18 and gender == "F":
    print("You are a woman")
else:
    print("You are a man")
```

```python
your_school = "Churchill High"
if your_school == "Churchill High" or your_school == "Hoover Middle" or your_school == "Beverley Farms Elementary":
    print("You are from my neighbood")
else:
    print("Welcome, stranger")
```

__Nested if statements__

```python
your_age = 14
your_gender = "F"

if your_age < 18:
    if your_gender == "F":
        print("You are a girl")
    else:
        print("You are a boy")
else:
    if your_gender == "F":
        print("You are a woman")
    else:
        print("You are a man")
```

__break and continue__

```python
# Condition "True" is always met in the while loop. We must use a "break" to terminate the loop.

secret_number = 14
while True:
    user_input = int(input("Try to guess the secret number from 0 to 20: "))
    if user_input == secret_number:
        print("Perfect, the secret number is", secret_number)
        break
    else:
        print("Wrong!")

print("You passed the test!")        
                     
```

```python
## "continue" means to skip the current iteration and move on to the next
for counter in range(1, 5):
    if counter == 3:
        continue
    print("counter:", counter)
```

__Other loop features__

```python
# When we develop a function and do not want to implement the for loop at this moment, we can simply put "pass" in the loop block. 
# The "pass" does nothing, but is only a placeholder. Without it, Python will throw an error.
for i in range(11):
    pass
```

```python
# The "while-else" loop will be executed at least once even if the while condition is never met.
i = 5
while i < 5:
    print(i)
    i += 1
else:
    print("else:", i)
```

### Section 3. Data Collections - Tuples, Dictionaries, Lists, and Strings

#### 3.1 List 

Lists are usually used to save multiple values of the same type (integers, strings, etc.)

```python
# Note that the elements in a list are indexed based on 0. The first element is indexed as "0", the second "1", so on. 
cities = ["New York", "Los Angeles", "Chicago", "Houston"]
print("The second biggest city in the US is:", cities[1])
```

```python
# A special notification is that "-1" means the last element in a list, "-2" means the second last, so on.
print("The last of the biggest cities is:", cities[-1])
```

```python
# "Slicing" of a list: to get part of the list by using indexes.
# Below example get the elements starting at 0 and stopping "BEFORE" 2 (2 is not included in the returned values).
print(cities[0:2])
```

```python
# Negative index can also be used in slicing. In below example, 
# note that "-1" is the last element and it is not included because the "end" index is exclusive.
print(cities[2:-1])
```

```python
# In slicing, if the start index is missing, it is default to "0", the beginning. 
# If end index is missing, it is default to include all elements from start index.
print(cities[:2])

# Below example is to print all elements in the list.
print(cities[:])
```

__Deleting elements from a list__

```python
us_cities = ["New York", "Los Angels", "Singapore", "Chicago", "Houston"]
# Singapore is error in this list, we can delete it.
del us_cities[2]

print(us_cities)
```

```python
us_cities = ["New York", "Los Angels", "Chicago", "Houston", "Phoenix"]
# We can use slicing to delete multiple elements. In below example, 1: "Los Angels", 2: "Chicago". Note that "end" index is exclusive.
del us_cities[1:3]

print(us_cities)
```

```python
# "del list name" means to delete the list itself. After this operation, the list becomes no longer available.
us_cities = ["New York", "Los Angels", "Chicago", "Houston", "Phoenix"]

del us_cities

print(us_cities)
```

__Adding elements to a list__

Note that "append" and "insert" are methods that belong to the list. A list calls these methods like ".append()". 

```python
us_cities = ["New York", "Los Angels", "Chicago", "Houston"]

us_cities.append("Phoenix")

print(us_cities)
```

```python
us_cities = ["New York", "Los Angels", "Chicago", "Houston"]

# Below example inserts "Phoenix" at index 3 of the list. All elements previously at and right of this index are shifted to the right.
us_cities.insert(3, "Phoenix")
print(us_cities)
```

__Iterating Lists__

We iterate a list in a way like "for ... in ...".

```python
us_cities = ["New York", "Los Angels", "Chicago", "Houston"]

for city in us_cities:
    print("Current city: ", city)
```

```python
# We can also iterate a list using indexes.
# Note the function "len()", which returns the length of the list.
for index in range(len(us_cities)):
    print("Current city: ", cities[index], ", Index: ", index)
```

__Changing Element Positions__

```python
first = "first"
second = "second"
# Below example swaps the values of the two variables.
first, second = second, first

print("first number is:", first)
print("second number is:", second)
```

```python
# We can use swaping to change the positions of elements in a list.
us_cities = ["New York", "Los Angels", "Chicago", "Houston"]

us_cities[0], us_cities[-1] = us_cities[-1], us_cities[0]
print(us_cities)
```

__Sorting a List__

```python
# Below example sorts the elements according to the first letter.
us_cities = ["New York", "Los Angels", "Chicago", "Houston"]
us_cities.sort()
print(us_cities)
```

```python
# The "sort()" method has a parameter to tell Python to sort elements in ascending or descening order. 
# The default order is ascending. To sort in descending order, we use ".sort(reverse=True)".
numbers = [1, 3, 5, 4, 2, 6]
numbers.sort(reverse = True)
print(numbers)
```

There is a function to sort a list, "sorted()". This is different from the method ".sort()". The latter belongs to the list, while the former is a function that belongs to Python itself.

```python
numbers = [1, 3, 5, 4, 2, 6]

sorted_numbers = sorted(numbers, reverse = True)

print("The original list 'numbers' is not changed: ", numbers)
print("Function sorted() returns a new list that is assigned to the variable 'sorted_numbers': ", sorted_numbers)
```
